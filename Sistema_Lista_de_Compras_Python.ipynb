```python
import csv

lista_compras = []

def carregar_csv():
    try:
        with open("/content/lista_compras.csv", mode='r', encoding='utf-8') as arquivo:
            leitor = csv.DictReader(arquivo)
            for linha in leitor:
                status = True if linha.get("Comprado") == "Sim" else False
                lista_compras.append({
                    "item": linha.get("Item", ""),
                    "categoria": linha.get("Categoria", "Geral"),
                    "quantidade": linha.get("Quantidade", "1"),
                    "comprado": status
                })
    except FileNotFoundError:
        pass

def salvar_csv():
    with open("/content/lista_compras.csv", mode='w', newline='', encoding='utf-8') as arquivo:
        escritor = csv.DictWriter(arquivo, fieldnames=["Item", "Categoria", "Quantidade", "Comprado"])
        escritor.writeheader()
        for item in lista_compras:
            escritor.writerow({
                "Item": item["item"],
                "Categoria": item.get("categoria", "Geral"),
                "Quantidade": item.get("quantidade", "1"),
                "Comprado": "Sim" if item["comprado"] else "Não"
            })

def main ():
  carregar_csv()
  print("Bem vindo a sua Lista de Compras!")
  while True:
    print("\nMenu:")
    print("1. Adicionar Nova Compra")
    print("2. Visualizar Compras")
    print("3. Marcar item como comprado")
    print("4. Remover item da Lista")
    print("5. Sair")
    escolha = input("Digite um número que corresponde a opção desejada:")

    if escolha == '1':
     item = input ("Digite o nome do item que deseja adicionar na Lista de Compras: ")
     adicionar_lista(item)
    elif escolha == '2':
     visualizar_lista ()
    elif escolha == '3':
     visualizar_lista()
     try:
       indice = int(input("Digite o número do item que deseja marcar como comprado: ")) -1
       marcar_comprado(indice)
     except ValueError:
       print("Entrada inválida. Por favor, digite um número válido.")
    elif escolha == '4':
     visualizar_lista()
     try:
       indice = int(input("Digite o número do item que deseja remover da lista de compras: ")) -1
       remover_item(indice)
     except ValueError:
       print("Entrada inválida. Por favor, digite um número válido.")
    elif escolha == '5':
     print ("Saindo da Lista de Compras ")
     break
    else:
     print("Opção inválida! Por favor, tente novamente.")

def adicionar_lista(item):
    # Insere no final da lista
    lista_compras.append({"item": item, "categoria": "Geral", "quantidade": "1", "comprado": False})
    salvar_csv()
    print(f"Item '{item}' adicionado com sucesso!")

def visualizar_lista():
    if not lista_compras:
        print("Nenhum item encontrado na lista de compras.")
        return
    print("\nLista de Compras:")
    for i, compra in enumerate(lista_compras):
        status = "[✓]" if compra["comprado"] else "[ ]"
        print(f"{i + 1}. {status} {compra['item']}")

def marcar_comprado(indice):
    if 0 <= indice < len(lista_compras):
        lista_compras[indice]["comprado"] = True
        salvar_csv()
        print(f"Item '{lista_compras[indice]['item']}' marcado como comprado!")
    else:
        print("Índice de item inválido.")

def remover_item(indice):
    if 0 <= indice < len(lista_compras):
        item = lista_compras.pop(indice)["item"]
        salvar_csv()
        print(f"Item '{item}' removido com sucesso!")
    else:
        print("Índice de item inválido.")

main()
```
